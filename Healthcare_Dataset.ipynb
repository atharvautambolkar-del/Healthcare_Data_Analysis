**Gender Count**

```python
gender_counts = df['Gender'].value_counts()
total_patients = len(df)
for gender, count in gender_counts.items():
    percentage = (count / total_patients) * 100
    print(f"  {gender}: {count} patients ({percentage:.1f}%)")
```

**Clean the Name columns to fix random uppercase/lowercase letters**

```python
df['Name'] = df['Name'].str.title()
df['Doctor'] = df['Doctor'].str.title()

print("Names cleaned!")
print("Patient name example:", df['Name'].iloc[0])
print("Doctor name example:", df['Doctor'].iloc[0])
```

```python
df.head()
```

**Convert Date columns to proper date format**

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

print("Date columns converted!")
print("Admission dates:", df['Date of Admission'].min(), "to", df['Date of Admission'].max())
```

**Calculate Length of Stay for each patient**

```python
df['Length_of_Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

print("Average stay:", round(df['Length_of_Stay'].mean(), 1), "days")
print("Shortest stay:", df['Length_of_Stay'].min(), "days")
print("Longest stay:", df['Length_of_Stay'].max(), "days")
```

**Chart: Most Common Diseases**

```python
disease_counts = df['Medical Condition'].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(disease_counts.index, disease_counts.values, color='skyblue')
plt.xlabel('Disease')
plt.ylabel('Number of Patients')
plt.title('Most Common Diseases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Chart: Gender Distribution**

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%', colors=['#66b3ff', '#ff9999'])
plt.title('Gender Distribution')
plt.show()
```

**Chart: Age Distribution**

```python
plt.figure(figsize=(8, 4))
plt.hist(df['Age'], bins=20, color='lightgreen', edgecolor='black')
plt.axvline(df['Age'].mean(), color='red', linestyle='--', label='Average Age')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.title('Age Distribution of Patients')
plt.legend()
plt.show()
```

**Chart: Most Prescribed Medications**

```python
med_counts = df['Medication'].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(med_counts.index, med_counts.values, color='#ffcc99')
plt.xlabel('Medication')
plt.ylabel('Number of Prescriptions')
plt.title('Most Prescribed Medications')
plt.tight_layout()
plt.show()
```
